```python
#2. Dot & cross products, and norms

# 2A. DOT & CROSS PRODUCTS

a = np.array([1, 2, 3])
b = np.array([4, 0, 1])

print('a . b  (dot)   :', np.dot(a, b))

print('a x b  (cross) :', np.cross(a, b))
```

```python

# 2B. NORMS (vector length) + COSINE SIMILARITY

print('L2 norm  ||a||_2 :', la.norm(a))
print('L1 norm  ||a||_1 :', la.norm(a, 2))
print('Linf norm        :', la.norm(a, np.inf))

def cosine(u, v):
    return np.dot(u, v) / (la.norm(u) * la.norm(v))

print('cosine(a, b)     :', round(cosine(a, b), 4))
```

```python
#LAB EXERCISE 2 — L1/L2 norms + cosine similarity on 3 vector pairs
print("Lab exercise 2- L1/L2 norms + cosine similarity on 3 vector pairs")
pairs = [
    (np.array([1, 0, 1]), np.array([1, 0, 1])),
    (np.array([1, 2, 3]), np.array([3, 2, 1])),
    (np.array([2, 0, 0]), np.array([0, 5, 0])),
]

similarities = []

for i, (u, v) in enumerate(pairs, start=1):

    l1 = np.linalg.norm(u, ord=1)
    l2 = np.linalg.norm(u)

    cosine = np.dot(u, v) / (np.linalg.norm(u) * np.linalg.norm(v))

    similarities.append(cosine)

    print(f"\nPair {i}:")
    print("L1 norm:", l1)
    print("L2 norm:", l2)
    print("Cosine similarity:", cosine)
    print()

most_similar = np.argmax(similarities) + 1
print("Most similar pair =", most_similar)
```

```python
#3. Matrix operations & special matrices

#3A. CORE MATRIX OPERATIONS

A = np.array([[2., 1.],
              [1., 3.]])
B = np.array([[1., 0.],
              [4., 2.]])

print('\nA @ B (matrix multiply):\n', A @ B)
print('\nA.T   (transpose):\n', A.T)
print('\ninverse(A):\n', la.inv(A))
print('\ntrace(A) = sum of diagonal:', np.trace(A))

```

```python

# 🔹 3B. SPECIAL MATRICES: identity, symmetric, orthogonal

I = np.eye(2)
print('\nIdentity I:\n', I)
print('\nA @ inv(A) == I ?', np.allclose(A @ la.inv(A), I))

print('\nA symmetric?', np.allclose(A, A.T))

theta = np.radians(30)
Q = np.array([[np.cos(theta), -np.sin(theta)],
              [np.sin(theta),  np.cos(theta)]])
print('\nQ orthogonal?', np.allclose(Q.T @ Q, I))
```

```python
# LAB EXERCISE 3 — Inverse, trace, and matrix tests
print("Lab exercise 3 - inverse , trace and matrix tests")
import numpy as np

M = np.array([[4., 2.],
              [2., 3.]])
P = np.array([[0., -1.],
              [1.,  0.]])

M_inv = np.linalg.inv(M)
trace_M = np.trace(M)

print("\nInverse of M:\n", M_inv)
print("\nTrace of M:", trace_M)

I = np.eye(2)
print("\nM @ inv(M):\n", M @ M_inv)
print("\nIs M @ inv(M) equal to I?", np.allclose(M @ M_inv, I))

print("\nIs M symmetric?", np.allclose(M, M.T))
print("\nIs P orthogonal?", np.allclose(P.T @ P, I))
```

```python
#4. Transformations: rotation & scaling

#  4A. BUILD TRANSFORMATION MATRICES

square = np.array([[0, 1, 1, 0],
                   [0, 0, 1, 1]], dtype=float)

S = np.array([[1.5, 0.0],
              [0.0, 0.5]])

t = np.radians(30)
R = np.array([[np.cos(t), -np.sin(t)],
              [np.sin(t),  np.cos(t)]])

scaled  = S @ square
rotated = R @ square
print('Rotated corners:\n', rotated)
```

```python
# 4B. PLOT ORIGINAL vs TRANSFORMED

def close_loop(pts):
    return np.hstack([pts, pts[:, :1]])

fig, ax = plt.subplots(figsize=(5, 5))
ax.plot(*close_loop(square),  marker='o', label='original')
ax.plot(*close_loop(scaled),  marker='o', label='scaled')
ax.plot(*close_loop(rotated), marker='o', label='rotated 30 deg')
ax.axhline(0, color='gray', lw=0.5); ax.axvline(0, color='gray', lw=0.5)
ax.set_aspect('equal'); ax.legend(); ax.set_title('Linear transformations')
```

```python
# LAB EXERCISE 4 — Apply 2 transformations & plot
print("Lab exercise 4 - apply 2 transformation and plot")
import numpy as np
import matplotlib.pyplot as plt

tri = np.array([[0, 1, 0.5],
                [0, 0, 1.0]])

S = np.array([[2.0, 0.0],
              [0.0, 0.5]])

theta = np.deg2rad(45)
R=np.array([[np.cos(theta), -np.sin(theta)],
              [np.sin(theta),  np.cos(theta)]])

transformed = R @ (S @ tri)

tri_plot = np.hstack((tri, tri[:, [0]]))
transformed_plot = np.hstack((transformed, transformed[:, [0]]))

plt.figure(figsize=(6, 6))
plt.plot(tri_plot[0], tri_plot[1], 'bo-', label='Original')
plt.plot(transformed_plot[0], transformed_plot[1], 'ro-', label='Scaled + Rotated')

plt.axis('equal')
plt.grid(True)
plt.legend()
plt.title("Triangle Transformation")
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```

```python
#5. Eigenvalues & eigenvectors


#  5A. COMPUTE EIGENVALUES & EIGENVECTORS

A = np.array([[2., 0.],
              [0., 3.]])

vals, vecs = la.eig(A)
print('\nEigenvalues  (lambda):', vals)
print('\nEigenvectors (columns):\n', vecs)
```

```python
#  5B. VERIFY  A v = lambda v

for i in range(len(vals)):
    v = vecs[:, i]
    lhs = A @ v
    rhs = vals[i] * v
    print(f'lambda={vals[i]:.1f}  A v == lambda v ?', np.allclose(lhs, rhs))
```

```python
# Eigenvalues/vectors & verification

print("Lab exercise 5- eigenvalues/vectors and verification")

C = np.array([[4., 1.],
              [2., 3.]])
eigenvalues, eigenvectors = np.linalg.eig(C)

print("\nEigenvalues:", eigenvalues)

v1 = eigenvectors[:, 0]
lambda1 = eigenvalues[0]

left_side = np.dot(C, v1)
right_side = lambda1 * v1

print("\nC @ v1:", left_side)
print("\nlambda1 * v1:", right_side)
print("\nEqual?", np.allclose(left_side, right_side))

```

```python
#6. Rank, solving systems & cosine similarity
#  6A. SOLVE A 3x3 SYSTEM  A x = b

A = np.array([[ 2.,  1., -1.],
              [-3., -1.,  2.],
              [-2.,  1.,  2.]])
b = np.array([8., -11., -3.])

x = la.solve(A, b)
print('Solution x:', x)
print('Rank of A :', la.matrix_rank(A))
print('Full rank -> a unique solution exists')
```

```python
# 6B. RANK TELLS YOU SOLVABILITY
D = np.array([[1., 2., 3.],
              [4., 5., 6.],
              [5., 7., 9.]])
print('Rank of D:', la.matrix_rank(D), '-> < 3, so rows are dependent')
```

```python
# LAB EXERCISE 6 — Solve a system + compare embeddings
print("Lab exercise 6- solve a system + compare embeddings\n")
A2 = np.array([[3., 2.], [1., 4.]])
b2 = np.array([7., 9.])

cat = np.array([0.9, 0.8, 0.1])
dog = np.array([0.85, 0.7, 0.2])
car = np.array([0.1, 0.2, 0.95])

x = np.linalg.solve(A2, b2)
print("1. Solution x =", x)

rank_A2=np.linalg.matrix_rank(A2)
print("2. Rank of A2 =", rank_A2)

def cosine_similarity(v1, v2):
    return np.dot(v1, v2) / (np.linalg.norm(v1) * np.linalg.norm(v2))

cos_cat_dog = cosine_similarity(cat, dog)
cos_cat_car = cosine_similarity(cat, car)

print(f"3. Cosine(cat, dog) = {cos_cat_dog:.4f}")
print(f"   Cosine(cat, car) = {cos_cat_car:.4f}")

if cos_cat_dog > cos_cat_car:
    print("   Result: 'cat' is more similar to 'dog'")
else:
    print("   Result: 'cat' is more similar to 'car'")
```
