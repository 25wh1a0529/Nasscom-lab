```python
from sklearn.preprocessing import StandardScaler,MinMaxScaler
data=([[25,50000],
       [35,100000],
       [45,150000]])
```

```python
standardized= StandardScaler().fit_transform(data)
print(standardized)
```

```python
normalized=MinMaxScaler().fit_transform(data)
print(normalized)
```
